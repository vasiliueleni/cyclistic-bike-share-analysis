```python
import numpy as np
import pandas as pd
from pathlib import Path 

DATA_PATH = Path("raw_data_csv")

q1_2019 = pd.read_csv(DATA_PATH / "Divvy_Trips_2019_Q1_csv.csv")
q1_2020 = pd.read_csv(DATA_PATH / "Divvy_Trips_2020_Q1_csv.csv")
```

# 2. Wrangle & Inspect Data

```python
# Compare column names for each of the files
print("Columns for Q1 2019:\n", q1_2019.columns)
print("\nColumns for Q1 2020:\n", q1_2020.columns)
```

```python
# Rename columns to make them consistent with q1_2020
q1_2019 = q1_2019.rename(columns={
    'trip_id':'ride_id',
    'bikeid':'rideable_type',
    'start_time':'started_at',
    'end_time':'ended_at',
    'from_station_name':'start_station_name',
    'to_station_name':'end_station_name',
    'from_station_id':'start_station_id',
    'to_station_id':'end_station_id',
    'usertype':'member_casual'
})
```

```python
# Inspect the dataframes to look for incongruities
q1_2019.info()
```

```python
q1_2020.info()
```

```python
# Convert ride_id and rideable_type to string so that they can stack correctly
q1_2019['ride_id'] = q1_2019['ride_id'].astype(str)
q1_2019['rideable_type'] = q1_2019['rideable_type'].astype(str)
```

```python
# Stack individual quarter's data frames into one big data frame
all_trips = pd.concat([q1_2019,q1_2020],ignore_index=True)
```

```python
all_trips.info()
```

```python
# Remove lat, long, birthyear & gender fields as this data was dropped in 2020
all_trips = all_trips.drop(columns=['start_lat','start_lng','end_lat',
                                    'end_lng','gender','birthyear',
                                    'tripduration'], 
                           errors='ignore')
```

```python
# Inspect the new table that has been created
print("\nList of column names:\n", all_trips.columns)
print("\nRows in the dataframe:", len(all_trips))
print("\nDimensions of the dataframe:", all_trips.shape)
```

```python
# Preview master dataframe
all_trips.head()
```

```python
all_trips.info()
```

# 3. Data Cleaning

```python
# In the 'member_casual' column, replace 'Subscriber' with 'member'
# and 'Customer' with 'casual'
# Begin by discovering how many observations fall under each user type
print("Value counts for 'member_casual' column before cleaning:")
all_trips['member_casual'].value_counts()
```

```python
# Reassign the desired values
all_trips['member_casual'] = all_trips['member_casual'].replace({
    'Subscriber':'member',
    'Customer':'casual'
})
```

```python
# Check to make sure the proper number of observations were reassigned
print("Value counts for 'member_casual' column after cleaning:")
all_trips['member_casual'].value_counts()
```

```python
# Convert 'started_at' and 'ended_at' to datetime objects
all_trips['started_at'] = pd.to_datetime(all_trips['started_at'])
all_trips['ended_at'] = pd.to_datetime(all_trips['ended_at'])
```

```python
# Add columns that list the date, month, day, and year of each ride 
all_trips['date'] = all_trips['started_at'].dt.date
all_trips['month'] = all_trips['started_at'].dt.month
all_trips['day'] = all_trips['started_at'].dt.day
all_trips['year'] = all_trips['started_at'].dt.year
all_trips['day_of_week'] = all_trips['started_at'].dt.day_name()
```

```python
# Add a 'ride_length' calculated field to all_trips (in seconds)
all_trips['ride_length'] = (all_trips['ended_at'] - 
                            all_trips['started_at']).dt.total_seconds()
```

```python
# Inspect the structure of the columns
print("Data types after adding date columns and ride_length:")
all_trips.info()
```

```python
# Remove "bad" data 
# The dataframe includes entries where bikes were taken out of docks for quality checks or ride_length was negative
# Create a new version of the data frame (v2) since data is being removed
all_trips_v2 = all_trips[(all_trips['start_station_name'] != "HQ QR") & 
    (all_trips['ride_length'] >= 0)].copy()
```

# 4. Descriptive Analysis

```python
# Set display format to 2 decimal places
pd.options.display.float_format = '{:.2f}'.format
```

```python
# Descriptive analysis on ride_length (all figures in seconds)
print("Descriptive statistics for ride_length:")
all_trips_v2['ride_length'].describe()
```

```python
# Compare members and casual users 
all_trips_v2.groupby('member_casual')['ride_length'].agg(['mean','median','max','min'])
```

```python
# See the average ride time by each day for members vs casual users 
all_trips_v2.groupby(['member_casual','day_of_week'])[['ride_length']].mean()
```

```python
# Order the days of the week for correct sorting analysis 
days_order = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
all_trips_v2['day_of_week'] = pd.Categorical(all_trips_v2['day_of_week'],
                                             categories=days_order, 
                                             ordered=True)
```

```python
# Now run the aggregation again to see the sorted result 
all_trips_v2.groupby(['member_casual', 'day_of_week'], observed=True)[['ride_length']].mean()
```

```python
# Analyze ridership data by type and weekday
summary_stats = all_trips_v2.groupby(['member_casual','day_of_week'], observed=True).agg(
    number_of_rides=('ride_id', 'count'),
    average_duration=('ride_length', 'mean')
).reset_index()

summary_stats
```
